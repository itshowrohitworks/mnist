# MNIST Dataset Exploring:

```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml("mnist_784",as_frame=False)
```

```python
X,y = mnist.data,mnist.target
```

```python
X
```

```python
y
```

There are 70000 Images and Each image has 28*28 Pixels (features)
Each feature represents one pixel's intensity from 0 (white) to 255 (black).

```python
X.shape,y.shape 
```

# Visual Representation of Image from the Data:

```python
import matplotlib.pyplot as plt
def plot_digit(data):
    image = data.reshape(28,28)
    plt.imshow(image,cmap="binary",interpolation="nearest",vmin=0, vmax=255)
    plt.axis("off")
```

```python
plot_digit(X[0])
plt.show()
```
